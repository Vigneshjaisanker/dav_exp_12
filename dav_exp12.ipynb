```python
# ============================================================
# Experiment: Independent Samples T-Test
# ============================================================

# Import Libraries
import pandas as pd
import numpy as np
from scipy.stats import ttest_ind
import os

# ------------------------------------------------------------
# Load the Datasets
# ------------------------------------------------------------

uci_file = "uci_diabetes (3).csv"
pima_file = "pima_diabetes (3).csv"

if os.path.exists(uci_file) and os.path.exists(pima_file):

    uci_diabetes = pd.read_csv(uci_file)
    pima_diabetes = pd.read_csv(pima_file)

    print("Both datasets loaded successfully!")

else:

    print("CSV files not found.")
    print("Creating sample datasets for demonstration...\n")

    np.random.seed(42)

    uci_diabetes = pd.DataFrame({
        "Glucose": np.random.normal(120, 25, 200),
        "BloodPressure": np.random.normal(72, 12, 200),
        "BMI": np.random.normal(32, 6, 200)
    })

    pima_diabetes = pd.DataFrame({
        "Glucose": np.random.normal(125, 28, 200),
        "BloodPressure": np.random.normal(70, 11, 200),
        "BMI": np.random.normal(31, 5, 200)
    })

# ------------------------------------------------------------
# Select Relevant Numerical Columns
# ------------------------------------------------------------

numerical_columns = [
    "Glucose",
    "BloodPressure",
    "BMI"
]

# ------------------------------------------------------------
# Perform Independent T-Test
# ------------------------------------------------------------

t_test_results = {}

for col in numerical_columns:

    # Remove missing values
    uci_values = uci_diabetes[col].dropna()
    pima_values = pima_diabetes[col].dropna()

    # Perform independent samples t-test
    t_stat, p_value = ttest_ind(
        uci_values,
        pima_values,
        equal_var=False
    )

    t_test_results[col] = {
        "T-statistic": t_stat,
        "P-value": p_value
    }

# ------------------------------------------------------------
# Convert Results to DataFrame
# ------------------------------------------------------------

t_test_df = pd.DataFrame(t_test_results).T

# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------

print("\nIndependent Samples T-Test Results")
print("-----------------------------------")

print(t_test_df.round(4))

# ------------------------------------------------------------
# Interpretation
# ------------------------------------------------------------

alpha = 0.05

print("\nInterpretation")
print("-----------------------------------")

for col in numerical_columns:

    p_value = t_test_df.loc[col, "P-value"]

    if p_value < alpha:
        print(
            f"{col}: Significant difference between the two datasets "
            f"(p = {p_value:.4f})."
        )
    else:
        print(
            f"{col}: No significant difference between the two datasets "
            f"(p = {p_value:.4f})."
        )

print("\nExperiment Completed Successfully!")
```
